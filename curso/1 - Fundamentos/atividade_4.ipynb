Crie um programa que responda se você foi aprovado numa prova. 
Você somente foi aprovado numa prova se sua média for maior ou igual que 7,
ou se sua nota no exame for maior ou igual a 5. Leia os valores do Input

```python
nota_prova = int(input("Digite o valor da nota de prova: "))
nota_exame = int(input("Digite o valor da nota de exame: "))

if nota_prova >= 7 or nota_exame >= 5:
    print("Voce esta aprovado")
else:
    print("Voce esta reprovado")
```

Crie  um programa que diga se a senha esta correta e portanto você tem acesso ao sistema. 
A senha devera ser salva no código, e a tentativa deve ser lida por input.

```python
senha = "123456"
senha_digitada = input("Digite uma senha: ")

print("Senha correta: ", senha_digitada == senha)
```
